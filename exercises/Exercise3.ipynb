## 3.1 Reading in data
***

You can write raw Python code to read in data files, but if coding up our `geo_sin()` function has taught us anything, it should be that we are much better off using a library to do basic work like this for us.

A powerful library for reading in data is `Pandas` - you can find documentation and handy examples __[here](https://pandas.pydata.org/docs/)__.  As before with NumPy, let's import the package and give it a short name that is convenient for typing out lots of times.

```python
import pandas as pd
```

You can read in a __[wide range of file types](https://pandas.pydata.org/docs/user_guide/io.html)__ with pandas.  One of the simplest file types is a 'comma-separated values' (CSV) file, which as the name suggests, just uses commas to separate columns of information in a file.  These are otherwise simple text files that you could open in many programs.

Let's open up a very simple data file just to see how it works.

```python
pd.read_csv('./data/mineral_top_trumps.csv', header=0, index_col=0)
```

If it has worked, a table should have appeared above.

How it worked:
- We are calling the `read_csv()` function from within pandas.
- The first argument to `read_csv()` is giving the location of the data file.
- The second argument, `header=0`, is saying that there is a header (i.e., column names) at the first row (remember that Python starts counting from zero).
- The third argument, `index_col=0`, is saying that there is an 'index column', i.e., the first column of the data file contains entries we want to use as the names of each row.

The last two arguments are **keyword arguments**: we give the name of the argument as well as its value.

It is not necessary to have either row names or column names, in which case they will just be assigned numerical values starting from 0.  But, how do we know whether the file has a header or row names before loading it in? 
>Always look at a data file before trying to read it in.

Let's select:
- Eruptions with names, &
- Eruptions with known volumes, &
- Eruptions with a composition listed for every element, &
- Eruptions from the Iceland Holocene (age < 12 kyr).

We will look closely at the first example, and then work quickly through the rest.

```python
#eruptions with names
# we can use the .notna() function to identify those entries with names listed
# i.e., those that are 'not a NaN'
df.Eruption.notna()

#this gives a lot of output... You can click on the cell edge to the left to collapse cells that you want to
# hide for convenience
```

```python
#now we need to use that list of true's and false's to actually select the data
# we can do this by passing the big list of true and false back to the dataframe
df[df.Eruption.notna()].head()

#we used head() again to just keep the output manageable, but you can see we 
# now just have eruptions with names.
```

```python
#Still, this isn't good enough, not all our eruptions have known volumes
# we can do the same as before and identify entries which don't have 'NaN' in the volume column
# but we now need to combine this with the selection for eruption name
# we use the '&' symbol for this i.e., we want eruptions with names _and_ with volumes
df[df.Eruption.notna() & df['Volume(km3)'].notna()].head()

#note, we had to use ['Volume(km3)'] rather than '.Volume(km3)' because of the parentheses
# in the column name.  Object methods, like functions and variables, can't be given names with parentheses 
# in them
```

That's two of our conditions met, maybe all these eruptions also have chemistry measured? Let's check.

```python
#first let's store our cleaned up data in a new dataframe
df_tidy = df[df.Eruption.notna() & df['Volume(km3)'].notna()]

#now, let's look to see if any of the oxides are missing in any of the eruptions
print( df_tidy.loc[:,'SiO2':'P2O5'].notna().all(axis=1) )

#How this works:
# we take df_tidy and we say we want to look at all rows, ':', and the major element data columns 'SiO2':'P2O5'
# now, we evaluate whether each entry is _not_ a NaN, i.e., we get a True value returned for every non-NaN
# now, using '.all()' we ask 'are all entries along the row = True', we specify to look along each row by passing 'axis=1' to '.all()'

# Every False value you see below is an eruption that doesn't have complete chemistry supplied.
```

```python
#finally, we take the list of true-false values and pass them back to df_tidy to select only complete chemical entries
df_Vtidy = df_tidy[df_tidy.loc[:,'SiO2':'P2O5'].notna().all(axis=1)]

#Look inside df_tidy at the row locations recording 'False' above
# to convince yourself that we really have found the entries with partial data
# Hint: use df_tidy.loc to do this
```

Lastly, we want to remove any eruptions older than 12 kyr

```python
#let's first look at the range of ages present to see whether there even are any old ones
df_Vtidy.age.describe()

#...yes there are!
```

```python
#we can filter numeric data very easily using standard comparison operators
# we should also make sure again that there are no NaN values in the ages
df_best = df_Vtidy[(df_Vtidy.age < 12000) & (df_Vtidy.age.notna())]
```

## 3.3 Operating on data frames
***

As we have seen hints of above, Pandas dataframes have a huge amount of inbuilt functionality (see just how much __[here](https://pandas.pydata.org/docs/reference/frame.html)__).  

First, let's answer the question we have all been waiting for.

So, Pandas can perform some simple maths for us on the data table.

Let's go back to the Icelandic eruption data and pull some statistics out of it, just using Pandas' built in functions.  (Note, that we could use NumPy functions to perform similar calculations, or even write our own).

```python
#first, I want to rename 'Volume(km3)' to something easier to type
# rename() gives us back a new dataframe with the column renamed, so we store it in df_best again
df_best = df_best.rename(columns={'Volume(km3)': 'volume'})

#Now, let's see what the average eruption volume is on Iceland
df_best.volume.mean()
```

```python
#how about comparing the eruption volumes of lavas with low MgO (<= 7.5 wt%) and with high MgO (> 7.5 wt%)
# we can use the conditional statements we learnt about previously to index the dataframe
vLmg = df_best[df_best.MgO <=7.50].volume.mean()

vHmg = df_best[df_best.MgO >7.50].volume.mean()

#let's print the output and use an f-string to reduce the output precision to something sensible
# it's surprisingly hard to measure the volume of volcanoes!
print(f'The average volume of high MgO eruptions is {vHmg:.2f} (km3)')

print(f'The average volume of low MgO eruptions is {vLmg:.2f} (km3)')
```

An interesting result!  The high MgO eruptions look smaller in eruptive volume than the low MgO eruptions. Is this difference meaningful? As a rough first look, let's compare the spread of the two populations, using their standard deviations.

```python
#first, if we are going to keep investigating these two subdatasets, it might be easier to create
# two new dataframes
dfLmg = df_best[df_best.MgO <= 7.5]
dfHmg = df_best[df_best.MgO > 7.5]

vstdLmg = dfLmg.volume.std()
vstdHmg = dfHmg.volume.std()

print(f'The average volume of high MgO eruptions is {vHmg:.2f} +/- {vstdHmg:.2f} (km3)\n')
print(f'The average volume of low MgO eruptions is {vLmg:.2f} +/- {vstdLmg:.2f} (km3)\n')


#let's write a simple if statement to check whether the two ranges of mean +/- 1 standard deviation overlap.
# if they do, the difference between the two means is probably not very meaningful.
# the ranges overlap if the bottom of the higher range is below the top of the lower range.

#even though we know vLmg > vHmg, sometimes it's good to write code that is robust to changes in the input
# like if we added more data
sad_outcome = ':( I\'m not going to be able to write a paper from this.'
good_outcome = 'Yay! Better get writing the Nature paper.'
if vLmg > vHmg:
    overlap = vLmg - vstdLmg < vHmg + vstdHmg
else:
    overlap = vHmg - vstdHmg < vLmg + vstdLmg

if overlap:
    print(sad_outcome)
else:
    print(good_outcome)
```

The ranges overlap, so the difference may not mean much. But the standard deviation tells us how spread out the individual values are, not how well we know each *mean*. A better tool for comparing two means is a **t-test**, which also takes account of how many values are in each group.

The SciPy library (a large collection of scientific tools, which we will use more in Exercise 5) has one:

```python
from scipy import stats

#Welch's t-test compares the means of two groups, without assuming they have the same spread
result = stats.ttest_ind(dfLmg.volume, dfHmg.volume, equal_var=False)

#the p-value is the probability of seeing a difference at least this large if the two groups
# really had the same mean. Conventionally, p < 0.05 is called 'significant'.
print(f'Comparing {len(dfLmg)} low MgO and {len(dfHmg)} high MgO analyses: p = {result.pvalue:.2g}')
```

A tiny p-value! Time to write that Nature paper after all?

Not so fast. Each row of `df_best` is a chemical *analysis*, not an eruption, and many eruptions have been analysed many times. The test above counts every analysis as a separate piece of evidence, so an eruption that has been analysed 50 times counts 50 times. That makes the result look far more certain than it really is.

To compare eruptions, we need one row per eruption. We can use `.groupby()` to collect together all the rows for each eruption, and then take the mean of each group. Some names are used for more than one eruption (for example, there are several lava flows of different ages called Burfellshraun), so we group by both the name and the age.

```python
#one row per eruption: group the analyses by eruption name and age, and take the mean of each group
df_erupt = df_best.groupby(['Eruption', 'age'], as_index=False).mean()

ve_Lmg = df_erupt[df_erupt.MgO <= 7.5].volume
ve_Hmg = df_erupt[df_erupt.MgO > 7.5].volume

result = stats.ttest_ind(ve_Lmg, ve_Hmg, equal_var=False)
print(f'Comparing {len(ve_Lmg)} low MgO and {len(ve_Hmg)} high MgO eruptions: p = {result.pvalue:.2g}')
```

Counting each eruption once, the p-value is right on the conventional borderline of 0.05, about a million times larger than before. There may be a difference, but the evidence is much weaker than the first test suggested.

Counting the same thing more than once is a very common mistake in data analysis, so always ask yourself what each row of your data represents.

## 3.4 Plotting<a id=sec14></a>
***

To plot data we are going to need to load a new library, `matplotlib`, specifically the plotting tools it contains, `pyplot`, which we access using the 'dot' syntax after `matplotlib`.  

```python
#again, we want to give our imported library a simple name that isn't too onerous to type out lots
import matplotlib.pyplot as plt
```

Now, there are a number of ways of producing plots with `matplotlib.pyplot`, offering different degrees of control and customisation options.  First, let's use the simplest method to plot some of the classic igneous differentiation diagrams from our dataset of Icelandic eruptions.

```python
#x data goes in first, followed by y data
plt.scatter(df_best.MgO, df_best.Na2O);

#note: I am putting ';' after the command to stop the object reference being printed
# try removing the semi-colon and seeing what gets printed
# although it seems useless, this output can be useful as we will see in a subsequent plot
```

```python
#let's now add some label names to our plot
# also, I'm not sure about these default colours, so let's change those
plt.scatter(df_best.MgO, df_best.Na2O,
            c='None',           #set colour of symbol centre, in this case an empty symbol
            edgecolor='black',  #set the edge colour to black
            linewidths=0.2);    #make the line width thin, this helps us see overlapping data points

# wt% = weight %, i.e., it is the oxide abundance by weight in the sample
plt.xlabel('MgO (wt%)')

#now, the y-label is a bit trickier, as really we want to write Na2O with a subscript on the 2
# we can do this using syntax from the LaTeX document writing language.  You may have heard of this
# and it's a great option for writing your Part III report.  However, for now, don't worry about the detailed syntax
plt.ylabel(r'Na${_2}$O (wt%)');
```

>As a reminder: Basalts start with high MgO and then evolve during crystallisation to low MgO as phases like olivine and clinopyroxene crystallise (you can see why these decrease MgO from their chemical formulae that we looked at above).  Now, whilst MgO is decreasing Na<sub>2</sub>O increases, because Na is mostly kept out of olivine and clinopyroxene and so becomes concentrated in the remaining melt (until plagioclase, which does take up Na, becomes abundant).

What would be interesting to know is whether the large volume eruptions are any more likely to lie at a particular point along this igneous differentiation trend.  To do this let's colour the plot symbols by the eruption volume.

```python
#Now add colour to the points, and make the symbols a bit larger so we can see them.
plt.scatter(df_best.MgO, df_best.Na2O, c=df_best.volume);

plt.xlabel('MgO (wt%)')
plt.ylabel(r'Na${_2}$O (wt%)')

#we should also add a colour bar
# note: in most (all?) programming languages spelling is American English!
plt.colorbar(label=r'Volume (km$^{3}$)');
```

Interestingly, it does look like large eruptions cluster at a very narrow range of MgO contents.  I wonder if anyone has published on this...

It would also be interesting to see how eruption volumes change through time. Let's try and put this as another panel on the same figure.  For this we are going to need to set up our plot in a slightly different way and we are going to want to plot a histogram, using `ax.hist(...)`.

```python
#first, let's create figure and axis objects, using 'subplots' to say that 
# we are going to be creating more than a single plot in this figure '.subplots(2)'
# says it is going to be two panels
fig, axs = plt.subplots(2)

#now we can plot like before, but with slightly different commands
# the plotting is actually a method contained by the axes object we just created
# as we have created more than one plot inside this figure, the axes need to be accessed
# using [] and indexes like we are used to for lists
im = axs[0].scatter(df_best.MgO, df_best.Na2O, c=df_best.volume);

#remember how previously we placed a ';' after some of the commands to stop them printing 
# out the reference to the object?  Well, now we need to save that information so that we can 
# refer to the specific plotting instance created by our command above 'axs[0].scatter(...'
# that is why the command starts 'im = ...'.
# we will use this to place a colour bar on the plot
fig.colorbar(im, ax=axs[0], label=r'Volume (km$^\mathsf{3}$)')

#we can set it up just like before, the commands are just a little different but self explanatory
axs[0].set_xlabel('MgO (wt%)')
axs[0].set_ylabel(r'Na${_2}$O (wt%)')
fig.set_size_inches(10, 10.5, forward=True)
#setting the physical size of the plot here so it is big enough to see

#now let's plot onto the bottom panel
# notice that we can do maths on the input if we want to transform units, for example.
# We also need to select only unique eruptions, i.e., we don't want to count an eruption
# multiple times just because there are multiple chemical analyses of it
# for this we use '.groupby(...)' again, just like we did for the t-test above
df_tmp = df_best.groupby(['Eruption', 'age'], as_index=False).mean()

#now we are ready to plot our histogram
n, bins, patches = axs[1].hist(df_tmp.age/1e3, weights=df_tmp.volume)

axs[1].set_xlabel('Age (kyr)')
axs[1].set_ylabel(r'Volume (km$^{3}$)')

#we need to call this to automatically move the axes around so there is 
# space for the labels.  Try commenting it out and replotting the figure
# to see what happens without it
fig.tight_layout()
```

Well, it looks like we got a lot of volume post-deglaciation at 12 kyr, and then there are some large eruptions more recently.  However, nearly 200 km<sup>3</sup> in the youngest bin looks a little large... Some of the eruptions in our data have an age of exactly 0, which is very unlikely: Thjórsárhraun, for example, is a famous lava flow that is about 8,600 years old. An age of 0 probably means 'unknown'. Try removing the eruptions with an age of 0 and plotting the histogram again.

We should probably worry about how good the age information we put in is, and read some of the __[relevant literature](https://agupubs.onlinelibrary.wiley.com/doi/abs/10.1029/2001GC000282)__ first!

### Saving your figure

Once you are happy with a figure you will want to save it, for example to put in a report.  The figure object has a `savefig()` method for this.  The type of file is set by the extension you give the file name:
- `.png` is an image made of pixels, so we also set its resolution with `dpi` (dots per inch). 300 is good for printing.
- `.pdf` and `.svg` are **vector** formats, which stay sharp however much you zoom in. These are usually the best choice for reports.

`bbox_inches='tight'` trims the empty space around the edge of the figure.  The new files appear in the same folder as this notebook: look for them in VS Code's Explorer panel.

You can save data in a similar way.  For example, a dataframe's `to_csv()` method writes it to a CSV file.

```python
#save the figure we just made, as an image and as a pdf
fig.savefig('iceland_eruptions.png', dpi=300, bbox_inches='tight')
fig.savefig('iceland_eruptions.pdf', bbox_inches='tight')

#save our filtered eruption data
df_best.to_csv('iceland_best.csv')
```
