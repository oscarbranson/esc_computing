## Have a go: cooling time of the oceanic lithosphere

We want $\tau = l^2 / (\pi^2\kappa)$ for $l$ = 100 km and $\kappa = 10^{-6}\,\mathsf{m}^2\,\mathsf{s}^{-1}$, converted to Myr.

````python
#we defined pi earlier in the exercise, but let's define it again so this cell works on its own
pi = 3.14159

#layer thickness (m) and thermal diffusivity (m^2/s)
l = 100e3
kappa = 1e-6

#cooling time in seconds
tau = l**2 / (pi**2 * kappa)

#convert to Myr, using the pi x 10^7 seconds in a year shortcut
tau_Myr = tau / (pi*1e7 * 1e6)

print('Cooling time of the oceanic lithosphere:', tau_Myr, 'Myr')
````

## Independent coding: predator prey interaction

> **Aim:** Write a while loop to calculate 100 years of predator prey interactions.

````python
#let's define our variables
# first for the trilobites
gx = 1.1 #growth
dx = 0.4 #death
x0 = 10  #initial population
# now for the nautiloids
gy = 0.1 #growth
dy = 0.4 #death
y0 = 10  #initial population

#set up the lists we are going to store all the results in
x = [x0]
y = [y0]
t = [0]

#time step
dt = 0.001
#maximum time to run our calculation until
tmax = 100

#start our loop
while t[-1] < tmax:
    #calculate both new populations from the current ones, x[-1] and y[-1], before storing either.
    # If we appended the new x first, x[-1] would already be the new value when we calculate y.
    x_new = x[-1] + (gx*x[-1] - dx*x[-1]*y[-1])*dt
    y_new = y[-1] + (gy*x[-1]*y[-1] - dy*y[-1])*dt

    #now store the new populations and the new time
    x.append(x_new)
    y.append(y_new)
    t.append(t[-1] + dt)

import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(t, x, label="Trilobites")
ax.plot(t, y, label="Nautiloids")
ax.set_xlabel('time (yr)')
ax.set_ylabel('population size')
ax.legend();

#save the figure to a file
fig.savefig('exercise1_predator_prey.png', dpi=300, bbox_inches='tight')
````
